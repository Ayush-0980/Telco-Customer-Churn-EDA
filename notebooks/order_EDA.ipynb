## Feature Engineering

```python
df['profit_per_unit'] = df['list_price'] - df['cost_price']
```

```python
df['discount_amount'] = df['list_price'] * df['discount_percent'] / 100
```

```python
df['selling_price'] = df['list_price'] - df['discount_amount']
```

```python
df['sales'] = df['selling_price'] * df['quantity']
```

```python
df['profit'] = df['profit_per_unit'] * df['quantity']
```

## Extract Date Features

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['day'] = df['order_date'].dt.day
df['month_name'] = df['order_date'].dt.month_name()
df['weekday'] = df['order_date'].dt.day_name()
```

## EDA Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Orders by Category

```python
sns.countplot(x='category', data=df)
plt.title("Number of Orders by Category")
plt.show()
```

### Sales by Region

```python
df.groupby('region')['sales'].sum().plot(kind='bar')
plt.title("Sales by Region")
plt.ylabel("Total Sales")
plt.show()
```

### Monthly Sales Trend

```python
import calendar

monthly_sales = df.groupby('month')['sales'].sum()

# Convert month numbers to month names
monthly_sales.index = monthly_sales.index.map(lambda x: calendar.month_name[x])

monthly_sales.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

### Profit by Category

```python
sns.barplot(x='category', y='profit', data=df)

plt.title("Profit by Category")
plt.show()
```

### Top 10 Products by Sales

```python
import matplotlib.pyplot as plt

top_products = df.groupby('product_id')['sales'].sum().sort_values(ascending=False).head(10)

colors = ['red','blue','green','orange','purple','cyan','magenta','yellow','brown','pink']

top_products.plot(kind='bar', color=colors)

plt.title("Top 10 Products by Sales")
plt.xlabel("Product ID")
plt.ylabel("Sales")

plt.show()
```
